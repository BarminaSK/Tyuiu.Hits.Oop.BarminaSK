```polyglot-notebook
using System;
using System.Collections.Generic;

public interface IInstalable
{
    void Install();
}

public abstract class Component
{
    public string Name { get; set; }
    public int Price { get; set; }

    public Component(string name, int price)
    {
        Name = name;
        Price = price;
    }

    public virtual string GetInfo()
    {
        return $"Название: {Name}, Цена: {Price}";
    }

    public override string ToString()
    {
        return GetInfo();
    } 
}

public class CPU : Component, IInstalable
{
    public int Cores { get; set; }
    public int Threads { get; set; }

    public CPU(string name, int price, int cores, int threads) : base(name, price)
    {
        Cores = cores;
        Threads = threads;
    }

    public override string GetInfo()
    {
        return $"Процессор: {Name}(Число ядер: {Cores}, Число потоков: {Threads}, Цена: {Price})";
    }

    public void Install()
    {
        Console.WriteLine($"Процессор {Name} установлен в сокет материнской платы");
    }
    
    public void UnIstall()
    {
        Console.WriteLine($"Процессор {Name} извлечен из сокета материнской платы");
    }
}

public class RAM : Component, IInstalable
{
    public int CapacityGB { get; set; }
    public int Type { get; set; }

    public RAM(string name, int price, int capacityGB, int type) : base(name, price)
    {
        CapacityGB = capacityGB;
        Type = type;
    }

    public override string GetInfo()
    {
        return $"ОЗУ: {Name}(Ёмкость: {CapacityGB}, Тип памяти: {Type}, Цена: {Price})";
    }

    public void Install()
    {
        Console.WriteLine($"ОЗУ {Name} установлено в слота DIMM");
    }
    
    public void UnIstall()
    {
        Console.WriteLine($"ОЗУ {Name} извлечено из слота DIMM");
    }
}

public class GPU : Component, IInstalable
{
    public int VramGB { get; set; }

    public GPU(string name, int price, int vramGB) : base(name, price)
    {
        VramGB = vramGB;
    }

    public override string GetInfo()
    {
        return $"Видеокарта: {Name}(Объём памяти: {VramGB}, Цена: {Price})";
    }

    public void Install()
    {
        Console.WriteLine($"Видеокарта {Name} установлен в слот PCI");
    }
    
    public void UnIstall()
    {
        Console.WriteLine($"Видеокарта {Name} извлечен из слота PCI");
    }
}

public class HDD : Component, IInstalable
{
    public int CapacityTB { get; set; }

    public HDD(string name, int price, int capacityTB) : base(name, price)
    {
        CapacityTB = capacityTB;
    }

    public override string GetInfo()
    {
        return $"Жёсткий диск: {Name}(Объём памяти: {CapacityTB}, Цена: {Price})";
    }

    public void Install()
    {
        Console.WriteLine($"Жёсткий диск {Name} установлен и подключен по SATA");
    }
    
    public void UnIstall()
    {
        Console.WriteLine($"Жётский диск {Name} извлечен из SATA");
    }
}

public class Computer
{
    private List<Component> _components = new List<Component>();

    public void AddComponent<T>(T component) where T : Component, IInstalable
    {
        _components.Add(component);
        component.Install();
    }

    public void RemoveComponent<T>(T component) where T : Component
    {
        _components.Remove(component);
    }

    public int GetTotalPrice()
    {
        int total = 0;
        foreach (var component in _components)
        {
            total += component.Price;
        }
        return total;
    }

    public void DisplayInfo()
    {
        foreach (var component in _components)
        {
            Console.WriteLine(component.GetInfo());
        }
        Console.WriteLine($"Общая стоимость: {GetTotalPrice()}");
    }
}

public static class ComputerFactory
{
    public static Computer CreateComputer<T_CPU, T_RAM, T_GPU, T_HDD>(T_CPU cpu, T_RAM ram, T_GPU gpu, T_HDD hdd)
        where T_CPU : CPU, IInstalable
        where T_RAM : RAM, IInstalable
        where T_GPU : GPU, IInstalable
        where T_HDD : HDD, IInstalable
    {
        Computer pc = new Computer();
        pc.AddComponent(cpu);
        pc.AddComponent(ram);
        pc.AddComponent(gpu);
        pc.AddComponent(hdd);
        return pc;
    }
}

Console.WriteLine("=== Сборка компьютера через Фабрику ===");

CPU intel = new CPU("Intel i5", 20000, 6, 12);
RAM kingston = new RAM("Kingston 16GB", 6000, 16, 5);
GPU rtx4060 = new GPU("RTX 4060", 40000, 8);
HDD wdBlue = new HDD("WD Blue 1TB", 5000, 1);

Computer myPC = ComputerFactory.CreateComputer(intel, kingston, rtx4060, wdBlue);

Console.WriteLine("\n=== Информация о созданном ПК ===");
myPC.DisplayInfo();

Console.WriteLine("\n=== Демонстрация удаления и добавления вручную ===");
myPC.RemoveComponent(wdBlue);
wdBlue.UnIstall();

Console.WriteLine("\n=== Информация о ПК после удаления диска ===");
myPC.DisplayInfo();
```
