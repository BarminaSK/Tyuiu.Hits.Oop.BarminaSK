### Описание задачи:
Создать базовый класс Person в C#, который будет представлять информацию о человеке. На основе этого класса разработать 2-3 производных класса, демонстрирующих принципы наследования и полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и методы, а также переопределены некоторые методы базового класса для демонстрации полиморфизма.
#### Требования к базовому классу Person:
* Атрибуты: Имя (Name), Возраст (Age), Пол (Gender).
* Методы:
	* GetInfo(): метод для получения информации о человеке в виде строки.
	* SayHello(): метод для вывода приветствия от человека.
Требования к производным классам:
1. Студент (Student): Должен содержать дополнительные атрибуты, такие как Университет (University) и Курс (Course). Метод SayHello() должен быть переопределен для добавления информации о курсе при приветствии.

2. Работник (Employee): Должен содержать дополнительные атрибуты, такие как Компания (Company) и Зарплата (Salary). Метод GetInfo() должен быть переопределен для включения информации о компании и зарплате.

3. Преподаватель (Teacher) (если требуется третий класс): Должен содержать дополнительные атрибуты, такие как Предмет (Subject) и Стаж работы (Experience). Метод SayHello() должен быть переопределен для добавления информации о преподаваемом предмете при приветствии.

#### Дополнительное задание
Добавьте к существующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте полиморфизм с перекрытием и перегрузкой методов, а также generic классы

```
using System;
using System.Collections.Generic;
```

```
public interface IResearcher
{
    void PublishArticle();
}

public interface ITaxPayer
{
    int TaxID { get; set; }
    bool DidYouPayTaxes { get; set; }
    void PayTaxes();
}
```

```
public class Person
{
    private string _name;
    private int _age;
    private string _gender;

    public string Name
    {
        get { return _name; }

        set
        {
            if (string.IsNullOrWhiteSpace(value))
            {
                throw new ArgumentException("Имя не может быть пустым");
            }
            else
            {
                _name = value;
            }
        }
    }

    public int Age
    {
        get { return _age; }

        set
        {
            if (value < 0)
            {
                throw new ArgumentException("Возраст не может быть отрицательным");
            }
            else
            {
                _age = value;
            }
        }
    }

    public string Gender
    {
        get { return _gender; }
        set { _gender = value; }
    }

    public Person(string name, int age, string gender)
    {
        Name = name;
        Age = age;
        Gender = gender;
    }

    public virtual string GetInfo()
    {
        return $"Имя: {Name}, Возраст: {Age}, Пол: {Gender}";
    }

    public virtual void SayHello()
    {
        Console.WriteLine($"Привет! Меня зовут {Name}.");
    }

    public void Greeting(Person anotherPerson)
    {
        Console.WriteLine($"{this.Name} поприветствовал {anotherPerson.Name}");
    }
    
    public void Greeting(Person anotherPerson, string question)
    {
        Console.WriteLine($"{this.Name} поприветствовал {anotherPerson.Name}. {question}?");
    }
}
```

```
public class Student : Person, IResearcher
{
    public string University { get; set; }
    public string Course { get; set; }
    public int Article { get; set; }
    public int Scholarship { get; set; }
    public double AverageGrade { get; set; }

    public Student(string name, int age, string gender, string university, string course, int article, int scholarship, double averageGrade)
    : base(name, age, gender)
    {
        University = university;
        Course = course;
        Article = article;
        Scholarship = scholarship;
        AverageGrade = averageGrade;
    }

    public override string GetInfo()
    {
        return $"Имя: {Name}, Возраст: {Age}, Пол: {Gender}, Университет: {University}, Курс: {Course}, Стипендия (в рублях): {Scholarship}, Средняя оценка: {AverageGrade}";
    }

    public override void SayHello()
    {
        Console.WriteLine($"Привет! Меня зовут {Name}, я студент {University}, {Course}.");
    }

    public void PublishArticle()
    {
        Console.WriteLine($"Количество статей, написанных студентом: {Article}");
    }

    public void PassExam(int grade)
    {
        if (grade == 5)
        {
            Scholarship += 1000;
            Console.WriteLine($"Стипендия студента {Name} увеличилась, теперь текущая стипендия: {Scholarship}");
        }
        else if (grade == 3 || grade == 4)
        {
            Console.WriteLine($"Стипендия студента {Name} осталась прежней: {Scholarship}");
        }
        else
        {
            Scholarship = 0;
            Console.WriteLine($"Студент {Name} лишен стипендии, стипендия: {Scholarship}");
        }
    }

}
```

```
public class Employee : Person, ITaxPayer
{
    public string Company { get; set; }
    public int Salary { get; set; }
    public int TaxID { get; set; }
    public bool DidYouPayTaxes { get; set; }

    public Employee (string name, int age, string gender, string company, int salary, int taxID, bool didYouPayTaxes)
    : base (name, age, gender)
    {
        Company = company;
        Salary = salary;
        TaxID = taxID;
        DidYouPayTaxes = didYouPayTaxes;
    }

    public override void SayHello()
    {
        Console.WriteLine($"Привет! Меня зовут {Name}, я работник {Company}");
    }


    public override string GetInfo()
    {
        return $"Имя: {Name}, Возраст: {Age}, Пол: {Gender}, Компания: {Company}, Зарплата: {Salary}";
    }

    public void PayTaxes()
    {
        Console.WriteLine($"ИНН сотрудника: {TaxID}, Налоги заплачены: {(DidYouPayTaxes ? "Да" : "Нет")}");
    }

    public void ChangeCompany(string newCompany)
    {
        Console.WriteLine($"{Name} теперь работает в {newCompany}");
    }
}
```

```
public class Teacher : Employee
{
    public string Subject { get; set; }
    public int Experience { get; set; }
    public string AssignedGroup { get; set; }
    public string AcademicTitle { get; set; }

    public Teacher (string name, int age, string gender, string company, int salary, int taxID, bool didYouPayTaxes, string subject, int experience, string assignedGroup, string academicTitle)
    : base (name, age, gender, company, salary, taxID, didYouPayTaxes)
    {
        Subject = subject;
        Experience = experience;
        AssignedGroup = assignedGroup;
        AcademicTitle = academicTitle;
    }

    public override void SayHello()
    {
        Console.WriteLine($"Привет! Меня зовут {Name}, я преподаю {Subject} и являюсь куратором группы {AssignedGroup}.");
    }

    public override string GetInfo()
    {
        return $"Имя: {Name}, Возраст: {Age}, Пол: {Gender}, Компания: {Company}, Зарплата: {Salary}, Предмет: {Subject}, Опыт (в годах): {Experience}, Ученое звание: {AcademicTitle}";
    }

    public void Lession(string currentGroup)
    {
        Console.WriteLine($"Преподаватель {Name} ведёт лекцию по {Subject} группе {currentGroup}");
    }

}
```

```
public class PersonCollections<T> where T: Person
{
    List<T> _people = new List<T>();

    public void Add(T person)
    {
        _people.Add(person);
    }

    public void Remove(T person)
    {
        _people.Remove(person);
    }

    public void DisplayInfo()
    {
        foreach(var person in _people)
        {
            Console.WriteLine(person.GetInfo());
        }
    }
}

PersonCollections<Person> university = new PersonCollections<Person>();

Student Ksyusha = new Student("Ксюша", 19, "Женский", "СПбГУ", "1 курс", 0, 7000, 3.6);

university.Add(new Student("Максим", 20, "Мужской", "МФТИ", "2 курс", 3, 10000, 4.9));
university.Add(Ksyusha);
university.Add(new Employee("Анна", 28, "Женский", "Магнит", 80000, 15844421, true));
university.Add(new Teacher("Елена", 35, "Женский", "ТИУ", 50000, 15944522, false, "Математика", 10, "ИСТНб-24-1", "Доцент"));

Console.WriteLine("--- Состав университета до удаления Ксюши ---\n");
university.DisplayInfo();
Console.WriteLine("\n--- Состав университета после удаления Ксюши ---\n");
university.Remove(Ksyusha);
university.DisplayInfo();
```

```
Student myStudent = new Student("Максим", 20, "Мужской", "МФТИ", "2 курс", 3, 10000, 4.9);
Employee myEmployee = new Employee("Анна", 28, "Женский", "Магнит", 80000, 15844421, true);
Teacher myTeacher = new Teacher("Елена", 35, "Женский", "ТИУ", 50000, 15944522, false, "Математика", 10, "ИСТНб-24-1", "Доцент");

List<Person> people = new List<Person>();
people.Add(myStudent);
people.Add(myEmployee);
people.Add(myTeacher);

Console.WriteLine("--- ДЕМОНСТРАЦИЯ ПОЛИМОРФИЗМА ---");

foreach(Person p in people)
{
    Console.WriteLine("\nИнфopмaция:");
    Console.WriteLine(p.GetInfo());
    Console.WriteLine("--- Знакомство ---");
    p.SayHello();
}
```
