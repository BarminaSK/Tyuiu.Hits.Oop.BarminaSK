<h2 style="color:DodgerBlue">Описание проекта:</h2>

Демонстрация наследования и полиморфизма в C# на примере иерархии классов Person, Student, Employee и Teacher.


<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
public class Person
{
    public string Name { get; set; }
    public int Age { get; set; }
    public string Gender { get; set; }

    public Person(string name, int age, string gender)
    {
        Name = name;
        Age = age;
        Gender = gender;
    }

    public virtual string GetInfo()
    {
        return $"Имя: {Name}, Возраст: {Age}, Пол: {Gender}";
    }

    public virtual void SayHello()
    {
        Console.WriteLine($"Привет! Меня зовут {Name}.");
    }
}
```

```
public class Student : Person
{
    public string University {get; set; }
    public string Course {get; set; }

    public Student(string name, int age, string gender, string university, string course)
    : base(name, age, gender)
    {
        University = university;
        Course = course;
    }

    public override void SayHello()
    {
        Console.WriteLine($"Привет! Меня зовут {Name}, я студент {University}, {Course}.");
    }

}
```

```
public class Employee : Person
{
    public string Company {get; set;}
    public int Salary {get; set;}

    public Employee (string name, int age, string gender, string company, int salary)
    : base (name, age, gender)
    {
        Company = company;
        Salary = salary;
    }

    public override string GetInfo()
    {
        return $"Имя: {Name}, Возраст: {Age}, Пол: {Gender}, Компания: {Company}, Зарплата: {Salary}";
    }
}
```

```
public class Teacher : Person
{
    public string Subject {get; set;}
    public int Experience {get; set;}

    public Teacher (string name, int age, string gender, string subject, int experience)
    : base (name, age, gender)
    {
        Subject = subject;
        Experience = experience;
    }

    public override void SayHello()
    {
        Console.WriteLine($"Привет! Меня зовут {Name}, я преподаю {Subject}.");
    }

}
```

```
List<Person> people = new List<Person>();

people.Add(new Student("Максим", 20, "Мужской", "МФТИ", "2 курс"));
people.Add(new Employee("Анна", 28, "Женский", "Магнит", 80000));
people.Add(new Teacher("Елена", 35, "Женский", "Математика", 10));

foreach (Person p in people)
{
    Console.WriteLine(p.GetInfo());
    p.SayHello();
}
```
