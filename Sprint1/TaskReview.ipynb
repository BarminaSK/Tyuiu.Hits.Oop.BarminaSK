```polyglot-notebook
public class Product
{
    private string name;
    private decimal price;
    private int quantity;
    protected string category;
    internal string description;


    public string Name
    {
        get {return name;}
        set {name = value;}
    }

    public int Quantity
    {
        get {return quantity;}
        set {quantity = value;}
    }

    public string Category
    {
        get {return category;}
        set {category = value;}
    }

    public string Description
    {
        get {return description;}
        set {description = value;}
    }

    public decimal Price
    {
        get {return price;}
        set {price = value;}
    }

    public decimal GetTotalPrice()
    {
        return price * quantity;
    }

    public void UpdateQuantity(int amount)
    {
        quantity += amount;
    }

    public string GetProductInfo()
    {
        return $"Товар: {name}, Цена: {price}, Количество: {quantity}, Категория: {category}, Описание: {description}";
    }

}
```

```polyglot-notebook
public class Seller
{
    private string name;
    private string employeeId;
    private decimal salary;
    internal string contactInfo;

    private List<Product> products = new List<Product>();

    public string Name
    {
        get {return name;}
        set {name = value;}
    }

    public string EmployeeId
    {
        get {return employeeId;}
        set {employeeId = value;}
    }

    public decimal Salary
    {
        get {return salary;}
        set {salary = value;}
    }

    public string ContactInfo
    {
        get {return contactInfo;}
        set {contactInfo = value;}
    }

    public void AddProduct(Product product)
    {
        products.Add(product);
    }

    public void SellProduct(Product product, int quantity)
    {
        if (products.Contains(product))
        {
            product.UpdateQuantity(-quantity);  
        }
    }

    public string GetSellerInfo()
    {
        return $"Продавец: {name}, ID: {employeeId}, Контакты: {contactInfo}";
    }
    
    public List<Product> GetProducts()
    {
        return products;
    }
}
```

```polyglot-notebook
public class Store
{
    private string storeName;
    private string location;
    private string storeHours;

    private List<Seller> sellers = new List<Seller>();

    public string StoreName
    {
        get {return storeName;}
        set {storeName = value;}
    }

    public string Location
    {
        get {return location;}
        set {location = value;}
    }

     public string StoreHours
    {
        get {return storeHours;}
        set {storeHours = value;}
    }

    public void AddSeller(Seller seller)
    {
        sellers.Add(seller);
    }

    public void ListProducts()
    {
        Console.WriteLine($"Список товаров в магазине: {storeName}:");

        foreach (var seller in sellers)
        {
            foreach (var product in seller.GetProducts())
            {
                Console.WriteLine(product.GetProductInfo());
            }
        }
    }

    public string GetStoreInfo()
    {
        return $"Магазин: {storeName}, Адрес: {location} , Часы работы: {storeHours}";
    }
    
}
```

```polyglot-notebook
Store myStore = new Store();
myStore.StoreName = "Техно-Мир";
myStore.Location = "ул. Главная, д. 10";
myStore.StoreHours = "09:00 - 21:00";
 
Seller sellerBob = new Seller();
sellerBob.Name = "Боб";
sellerBob.EmployeeId = "EMP001";
sellerBob.Salary = 50000;
sellerBob.ContactInfo = "bob@store.com";

Product laptop = new Product();
laptop.Name = "Ноутбук";
laptop.Price = 75000;
laptop.Quantity = 10;
laptop.Category = "Электроника";
laptop.Description = "Игровой ноутбук 16 ГБ ОЗУ";
 
sellerBob.AddProduct(laptop);
myStore.AddSeller(sellerBob);
 
Console.WriteLine(myStore.GetStoreInfo());
Console.WriteLine(sellerBob.GetSellerInfo());
Console.WriteLine();
 
Console.WriteLine("До продажи");
myStore.ListProducts();
Console.WriteLine($"Общая стоимость партии ноутбуков: {laptop.GetTotalPrice()} руб.");
Console.WriteLine();
 
Console.WriteLine("Боб продал 3 ноутбука");
sellerBob.SellProduct(laptop, 3);
Console.WriteLine();
 
Console.WriteLine("После продажи");
myStore.ListProducts();
```
